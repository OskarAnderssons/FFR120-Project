```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ast
def sum_of_consecutive_differences(points):
    sum_diff = 0
    for i in range(1, len(points)):
        sum_diff += abs(points[i] - points[i-1])
    return sum_diff
# Load the CSV files
df_fish_05 = pd.read_csv("fish_05_2.csv")
df_shark_05 = pd.read_csv("shark_05_2.csv")

# Convert the string representations of lists to actual Python lists
df_fish_05['X Positions'] = df_fish_05['X Positions'].apply(ast.literal_eval)
df_fish_05['Y Positions'] = df_fish_05['Y Positions'].apply(ast.literal_eval)
df_shark_05['X Positions'] = df_shark_05['X Positions'].apply(ast.literal_eval)
df_shark_05['Y Positions'] = df_shark_05['Y Positions'].apply(ast.literal_eval)

# Access X and Y data as lists and halve the data by taking every other point
fish_x_05 = df_fish_05['X Positions'][0][::2]  # Take every other point for Fish X
fish_y_05 = df_fish_05['Y Positions'][0][::2]  # Take every other point for Fish Y
shark_x_05 = df_shark_05['X Positions'][0][::2]  # Take every other point for Shark X
shark_y_05 = df_shark_05['Y Positions'][0][::2]  # Take every other point for Shark Y

# Create alpha values using an exponential function
n_points = len(fish_x_05[0:-32])
fish_alpha = 1 - np.exp(-0.08 * np.arange(n_points))  # Exponential decay for fish

n_points_shark = len(shark_x_05[1:-31])
shark_alpha = 1 - np.exp(-0.08 * np.arange(n_points_shark))  # Exponential decay for shark

# Plot scatter points for fish with exponentially varying transparency
for i in range(n_points):
    if i == len(fish_x_05[0:-33]):
        plt.scatter(fish_x_05[i], fish_y_05[i], color="blue", alpha=fish_alpha[i], label = "Sardine")
    # elif i ==3:
    #     plt.scatter(fish_x_05[i], fish_y_05[i], color="blue", alpha=fish_alpha[i], label = "sardine, first time step")
    else:
        plt.scatter(fish_x_05[i], fish_y_05[i], color="blue", alpha=fish_alpha[i])

# Plot scatter points for shark with exponentially varying transparency
for i in range(n_points_shark):
    if i == len(shark_x_05[1:-32]): 
        plt.scatter(shark_x_05[i], shark_y_05[i], color="red", alpha=shark_alpha[i], label = "Marlin")
    # elif i == 3:
    #     plt.scatter(shark_x_05[i], shark_y_05[i], color="red", alpha=shark_alpha[i], label = "Marlin, first time step")
    else:
        plt.scatter(shark_x_05[i], shark_y_05[i], color="red", alpha=shark_alpha[i])

print(sum_of_consecutive_differences(shark_x_05[1:-31]))
# Show plot
plt.xlabel("x [m]", fontsize=18)
plt.ylabel("y [m]", fontsize=18)
plt.xticks(fontsize = 14)
plt.yticks(fontsize = 14)
plt.legend(loc = 6, fontsize = 14)
plt.title("Marlin Chasing Sardine, -0.5s Delay", fontsize = 20)
plt.xlim(0, 75)
plt.ylim(0, 75)

plt.show()
```
